```python
import scipy.stats as stats
import statsmodels.stats.api as sms
import pandas as pd
import numpy as np
from math import ceil, sqrt
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
```

```python
test = pd.DataFrame(data = [['Control',1000,500],['Test',2500,780]],columns = ['Group','Visit','Conversion'],)
```

```python
test['CVR'] = test['Conversion']/test['Visit']
```

```python
(test.loc[1,'CVR',]/test.loc[0,'CVR',])-1
```

```python
effect_size = sms.proportion_effectsize(test.loc[0,'CVR'],test.loc[1,'CVR'])
required_n = sms.NormalIndPower().solve_power(effect_size, power = 0.8, ratio = 1, alpha=0.5)
```

```python
print(effect_size,'\n',ceil(required_n))
```

```python
nobs = [test.loc[0,'Visit'],test.loc[1,'Visit']]
success = test['Conversion']
z_stats, p_val = proportions_ztest(success,nobs=nobs)
```

```python
print(round(z_stats,ndigits = 4))
print('%.3f'%p_val)
```

```python
tail = stats.norm.sf(0.05)*2
print('%.5f'%tail)
```

```python
(lower_con,lower_test), (upper_con,upper_test) = proportion_confint(success,nobs = nobs)
```

```python
proportion_confint(success,nobs = nobs)
```

```python
print(lower_con,'\n',lower_test)
```

```python
print(f'z statistic: {z_stats:.2f}')
print(f'p-value: {p_val:.3f}')
```
