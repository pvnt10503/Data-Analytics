```python
edgeService = msedgedriver.install()
options = webdriver.EdgeOptions()
options.add_argument("--auto-open-devtools-for-tabs")
options.add_argument('--headless')
```

```python
driver = webdriver.Edge(service= edgeService,options=options)
driver.get('https://trends.google.com/trending?geo=VN&hl=vi&hours=168&sort=recency')
sleep(random.randint(5,10))
```

```python
count = 1
button_selector = '/html/body/c-wiz/div/div[5]/div[1]/c-wiz/div/div[2]/div[1]/div[1]/div[2]/div/div[2]/span[3]/button/span[2]'
title, num_searches, hours = [], [], []
while count <= max_page:
    try:
        print('Crawl Page '+ str(count))
        #crawl data in the current page
        elems = driver.find_elements(By.CSS_SELECTOR, '.mZ3RIc')
        title.extend(elem.text for elem in elems)
        sleep(3)
        elems_num_searches = driver.find_elements(By.CSS_SELECTOR, '.qNpYPd')
        num_searches.extend(num.text for num in elems_num_searches)
        elems_activate_hours = driver.find_elements(By.CSS_SELECTOR, '.A7jE4')
        hours.extend(hour.text for hour in elems_activate_hours)
        #click next page
        button = driver.find_element('xpath', button_selector)
        driver.execute_script("arguments[0].click();", button)
        print('Clicked on button next page!')
        sleep(random.randint(5,10))
        count += 1;
    except ElementNotInteractableException:
        print('Element Not Interactable Exception!') 
        break
```

```python
df1 = pd.DataFrame(list(zip(title,num_searches,hours)), columns = ['title','num_searches','activate_hours'])
```

```python
for i in range(len(df1)):
    df1['num_searches'][i]= df1['num_searches'][i][0:4].strip().replace(' N','000').replace('Tr','000000').replace('+','').replace(' ','')
df1['num_searches'] = df1['num_searches'].astype(float)
```

```python
df1.rename(columns = {'activate_hours':'start_hours'},inplace=True)
```
