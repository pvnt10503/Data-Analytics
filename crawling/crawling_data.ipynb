```python
def convert_sales(df,col):
    df[col] = df[col].replace('N/A',None)
    for i in range(0,len(df1)+1):
        try:
            df[col][i] = df[col][i].replace('.','.').replace('m','')
        except: 
            df[col][i] = None
    df[col]= df[col].astype(float)
    return df
#total_shipped is in Million
```

```python
convert_sales(df1,'total_shipped')
convert_sales(df1,'total_sales')
convert_sales(df1,'NA_sales')
convert_sales(df1,'PAL_sales')
convert_sales(df1,'Japan_sales')
convert_sales(df1,'other_sales')
```

```python
df1['critic_score'] = df1['critic_score'].replace('N/A',None)
df1['critic_score'] = df1['critic_score'].astype(float)
```

```python
#convert object to datetime:
def date_time(df,col):
    df[col] = df[col].replace('N/A',None)
    df[col] = pd.to_datetime(df[col]).dt.strftime('%Y-%m-%d')
    return df[col]
```

```python
date_time(df1,'released_date')
date_time(df1,'last_updated')
```

```python
df1['released_date'] = pd.to_datetime(df1['released_date'], yearfirst= True, format='%Y-%m-%d')
df1['last_updated'] = pd.to_datetime(df1['last_updated'], yearfirst= True, format='%Y-%m-%d')
```

```python
df1.info()
```

```python
df1['current_day'] = datetime.now().strftime('%Y-%m-%d')
df1['current_day'] = pd.to_datetime(df1['current_day'],format = '%Y-%m-%d')
df1.iloc[50]
df1['released_date'][50] = '2017-12-03'
df1['existing_time'] = (df1['current_day'] - df1['released_date']).dt.days
df1.drop(columns = 'current_day', inplace = True)
```

```python
df1['days_since_last_update'] = (df1['last_updated'] - df1['released_date']).dt.days
```

```python
swap_index = df1[df1['days_since_last_update']< 0].index
for index in swap_index:
    df1['released_date'][index], df1['last_updated'][index] = df1['last_updated'][index], df1['released_date'][index]
df1['days_since_last_update'] = (df1['last_updated'] - df1['released_date']).dt.days
```
