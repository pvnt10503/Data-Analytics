```python
import pandas as pd
import numpy as np
from selenium import webdriver
from selenium.webdriver.edge.service import Service
from selenium.webdriver.common.by import By
from time import sleep
import random
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.support.ui import WebDriverWait
import msedgedriver
from selenium.common.exceptions import JavascriptException, NoSuchElementException
from selenium.webdriver import DesiredCapabilities
```

```python
edgeService = msedgedriver.install()
options = webdriver.EdgeOptions()
options.add_argument("--auto-open-devtools-for-tabs")
options.add_argument('start-maximized')
options.add_argument('--headless')
options.add_argument('inprivate')
```

```python
driver = webdriver.Edge(service=edgeService, options= options)
driver.get('https://www.vgchartz.com/games/games.php?page=1&order=Sales&ownership=Both&direction=DESC&showtotalsales=1&shownasales=1&showpalsales=1&showjapansales=1&showothersales=1&showpublisher=1&showdeveloper=0&showreleasedate=1&showlastupdate=1&showvgchartzscore=0&showcriticscore=1&showuserscore=0&showshipped=1')
sleep(random.randint(5,10))
```

```python
games_titles, games_links, publishers, total_shipped, released_date, last_updated = [], [], [], [],[],[]
```

```python
count = 1
pages = 13
while count <= pages:
    try:
        for i in range(4,54):
            n = 3
            while n <= 8:
                xpath = f'//*[@id="generalBody"]/table[1]/tbody/tr[{i}]/td[%d]/a[1]'%n
                elems = driver.find_element(By.XPATH,xpath)
                games_titles.append(elems.text) 
                games_links.append(elems.get_attribute('href'))
                n +=2
                elems = driver.find_element(By.XPATH,f'//*[@id="generalBody"]/table[1]/tbody/tr[{i}]/td[%d]'%n)
                publishers.append(elems.text)
                n +=2
                elems = driver.find_element(By.XPATH,f'//*[@id="generalBody"]/table[1]/tbody/tr[{i}]/td[%d]'%n)
                total_shipped.append(elems.text)
                n += 6
                elems = driver.find_element(By.XPATH,f'//*[@id="generalBody"]/table[1]/tbody/tr[{i}]/td[%d]'%n)
                released_date.append(elems.text)
                n +=1
                elems = driver.find_element(By.XPATH,f'//*[@id="generalBody"]/table[1]/tbody/tr[{i}]/td[%d]'%n)
                last_updated.append(elems.text)
        print(f'Crawling data page {count}')
        count +=1
        next_page = driver.find_element(By.XPATH,'//*[@id="generalBody"]/table[2]/tbody/tr/th[2]/span[%d]/a'%count)
        driver.execute_script("arguments[0].click();", next_page)
        print(f'Clicked on button page {count}!')
        sleep(random.randint(5,10))
    except: NoSuchElementException
```

```python
df1 = pd.DataFrame(list(zip(games_titles,games_links, publishers, total_shipped, released_date, last_updated)),
                   columns = ['games_titles','games_links','publishers','total_shipped','released_date','last_updated'])
```
